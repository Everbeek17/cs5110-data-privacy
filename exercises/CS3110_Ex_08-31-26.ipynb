## Question 4

Conduct a linking attack to recover Brenn's data from the `adult_deid` dataset.


```python
# We have zip and age available, does this uniquely identify Brenn? Let's check:
```

```python
def recover_brenns_row():
    brenn_age = 38
    brenn_zip = 95668

    # NOTE: in pandas you need to use single &, not double &&. And you
    # need to put parentheses around each condition.
    return adult[(adult["Age"] == brenn_age) & (adult["Zip"] == brenn_zip)]


brenns_recovered_row = recover_brenns_row()
```

```python
assert len(brenns_recovered_row) == 1
assert brenns_recovered_row["Education"].iloc[0] == "HS-grad"
```
