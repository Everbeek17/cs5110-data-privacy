```python
# Load the data and libraries
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt


def laplace_mech(v, sensitivity, epsilon):
    return v + np.random.laplace(loc=0, scale=sensitivity / epsilon)


# Sensitivity parameter must be L1 sensitivity of the vector.
def laplace_mech_vec(vec, sensitivity, epsilon):
    return [
        v + np.random.laplace(loc=0, scale=sensitivity / epsilon) for v in vec
    ]


def gaussian_mech(v, sensitivity, epsilon, delta):
    return v + np.random.normal(
        loc=0, scale=sensitivity * np.sqrt(2 * np.log(1.25 / delta)) / epsilon
    )


# Sensitivity parameter must be L2 sensitivity of the vector.
# (It could actually be L1 sensitivity as well, but L1 is worse)
def gaussian_mech_vec(vec, sensitivity, epsilon, delta):
    return [
        v
        + np.random.normal(
            loc=0,
            scale=sensitivity * np.sqrt(2 * np.log(1.25 / delta)) / epsilon,
        )
        for v in vec
    ]


def pct_error(orig, priv):
    return np.abs(orig - priv) / orig * 100.0


adult = pd.read_csv(
    "https://github.com/jnear/cs3110-data-privacy/raw/main/homework/adult_with_pii.csv"
)
retail = pd.read_csv(
    "https://github.com/jnear/cs3110-data-privacy/raw/main/homework/retail.csv"
)
```

## Question 1

Implement the Gaussian mechanism for $(\epsilon, \delta)$-differential privacy.


```python
def gaussian_mech(v, sensitivity, epsilon, delta):
    # laplace did: return v + np.random.laplace(loc = 0,
    # scale=sensitivity/epsilon)

    # Why are there constants? This is not the best way to do this.
    # Why do we have a log() part and a delta? tbd
    # What is delta and how do we set it? It's a "catastrophic failure"
    # probability. So we want it as small as possible.
    sigma_squared = 2 * sensitivity**2 * np.log(1.25 / delta) / (epsilon**2)
    return v + np.random.normal(loc=0, scale=np.sqrt(sigma_squared))


gaussian_mech(len(adult), sensitivity=1, epsilon=1.0, delta=1e-5)
```

```python
# TEST CASE

results = [
    gaussian_mech(len(adult[adult["Age"] > 50]), 1, 1.0, 10e-5)
    for _ in range(100)
]
errors = [pct_error(len(adult[adult["Age"] > 50]), r) for r in results]
print("mean error:", np.mean(errors))

assert np.mean(errors) > 0
assert np.mean(errors) < 2
```

## Question 2

How do the Laplace and Gaussian mechanisms compare in terms of relative error on the query "how many individuals are over 50 years old" with $\epsilon = 1$ and $\delta = 10^{-5}$?


```python
true_answer = len(adult[adult["Age"] > 50])

laplace_answers = [laplace_mech(true_answer, 1, 1) for _ in range(200)]
gaussian_answers = [gaussian_mech(true_answer, 1, 1, 10e-5) for _ in range(200)]

laplace_error = [pct_error(true_answer, a) for a in laplace_answers]
gaussian_error = [pct_error(true_answer, a) for a in gaussian_answers]

_, bins, _ = plt.hist(gaussian_error, bins=20, label="Gaussian")
plt.hist(laplace_error, bins=bins, label="Laplace", alpha=0.5)
plt.legend()
```

The Laplace mechanism generally gives more accurate answers for this query than
the Gaussian mechanism does, even if we use the same value of epsilon.

## Question 3

The function `age_cdf` calculates the cumulative distribution function (CDF) of the ages in the adult dataset, expressed as a vector of counts.

What is the **L1 global sensitivity** of `age_cdf`?


```python
def age_cdf():
    a = adult["Age"]
    return [len(a[a < i]) for i in range(100)]
    # Length is 100
    # each element has sensitivity 1 (because they are all counting queries)
    # The L1 sensitivity is therefore 1+1+...+1 (100 times) = 100


plt.plot(age_cdf())
print("Length of CDF vector:", len(age_cdf()))
# NB: sensitivity is sum of sensitivities, not the sum of the values themselves
```

The L1 sensitivity is 100, because each element is a counting query with
sensitivity 1, and there are 100 elements.

## Question 5

Use the vector-valued Laplace and Gaussian mechanisms to construct noisy versions of the CDF of ages. **Be sure to use the right sensitivity!**


```python
def laplace_cdf(epsilon):
    # Use laplace_mech_vec with L1 sens
    l1_sens = 100
    return laplace_mech_vec(age_cdf(), sensitivity=l1_sens, epsilon=epsilon)


def gaussian_cdf(epsilon, delta):
    # use gaussian_mech_vec with L2 sens
    l2_sens = np.sqrt(100 * 1**2)  # = 10
    return gaussian_mech_vec(
        age_cdf(), sensitivity=l2_sens, epsilon=epsilon, delta=delta
    )


plt.plot(laplace_cdf(0.1), label="Laplace L1")
plt.plot(gaussian_cdf(0.1, 1e-5), label="Gaussian L2")
plt.legend()
```

When is L2 << L1?
It is NOT when the vector is very short, in short vector scenario L1 and L2 are
very similar. When vector is 1 element and counting query: L1 sens = L2 sens

When the vector is **long**, then L2 can be massively less than L1.
In our example above, L1 was n*1, and L2 was sqrt(n*(1^2)).

L1 sensitivity grows linearly with the length of the vector.
L2 sensitivity grows with its square root, which is better!
