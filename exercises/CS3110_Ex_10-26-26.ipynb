## Question -2

Write a function to calculate the L1 **local sensitivity** of the following mean query.


```python
def mean_query():
    b = 125
    return (
        adult["Age"][adult["Education"] == "Bachelors"]
        .clip(lower=0, upper=b)
        .mean()
    )


mean_query()
```

```python
def mean_query_local_sensitivity():
    b = 125
    numerator = b
    denominator = len(adult["Age"][adult["Education"] == "Bachelors"]) - 1
    # reason for plus one is because the neighboring dataset could have one
    # fewer row.
    return numerator / denominator


mean_query_local_sensitivity()
```

## Question 2

Implement `report_noisy_max` using the Laplace mechanism. `report_noisy_max` should return the value in a set that approximately maximizes the value of the score function. It should _not_ return the score itself.


```python
# NB: this is a common special case of the exponential mechanism for finite sets.
# This is a recurring function (report_noisy_max).


def report_noisy_max(R, score, sensitivity, epsilon):
    scores = [score(r) for r in R]
    # then we add noise
    noisy_scores = [
        laplace_mech(s, sensitivity=sensitivity, epsilon=epsilon)
        for s in scores
    ]

    # get the index of the item with the largest score
    max_idx = np.argmax(noisy_scores)
    # return the item within R that had the maximum noisy score
    return R[max_idx]


report_noisy_max(options, score, 1, 1)
```

```python
# TEST CASE
assert report_noisy_max(options, score, 1, 1) == "Married-civ-spouse"
```

## Question 3

What is the **total privacy cost** of `report_noisy_max` under _sequential composition_?


1. Sensitivity: score function is a countirng query, so the sensitivity for the
   laplace mech is 1.
2. Composition: we run the laplace mech "len(R)" times, so the total privacy
   cost is "len(R) \* epsilon" by sequential composition.
3. Post-Processing: "max_idx" is a function of the noisy scores (so it satisfies
   DP by post-processing) and "R[Max_idx]" is a function of "max_idx" (which
   satisfies DP) and "R" (which is public).

**BUT**: There is a special proof that says the total privacy cost of "report
noisy max" is _just_ epsilon.


The special proof takes advantage of the fact that "report noisy "max" returns
**only the identity of the best item** - it does _not_ return any info about the
other items, or any info about the scores. Hiding this info is what gives the
improved privacy budget.


## Above Threshold

The following code implements Above Threshold:


```python
# preserves epsilon-differential privacy
def above_threshold(queries, df, T, epsilon):
    T_hat = T + np.random.laplace(loc=0, scale=2 / epsilon)

    for idx, q in enumerate(queries):
        nu_i = np.random.laplace(loc=0, scale=4 / epsilon)
        if q(df) + nu_i >= T_hat:
            return idx
    return -1  # the index of the last element
```

## Question 4

Use `above_threshold` to find the first `age` for which `len(adult[adult['Age'] == age]) >= 800` (the first age for which more than 800 people have that age).


```python
def make_query(age):
    def q(df):
        return len(df[df["Age"] == age])

    return q


# This is returning the index of the query that first passes the threshold
# i.e. the first age with count above 800
def find_first_age_above_800(epsilon):
    all_queries = [make_query(age) for age in range(1, 100)]
    return above_threshold(all_queries, adult, T=800, epsilon=epsilon)


find_first_age_above_800(1.0)
```

## Question 5

Use `above_threshold` to implement `pick_b` for the following summation query. `pick_b` should pick a clipping parameter `b`.


```python
def make_query(b):
    def q(df):
        # return a positive value if I have reached the "plateau" of the b
        # graph.
        sum1 = df["Age"].clip(lower=0, upper=b).sum()
        sum2 = df["Age"].clip(lower=0, upper=b - 1).sum()
        return sum2 - sum1
        # The sensitivity here is 1, since changing one row changes sum by at
        # most 1. Not easily seen (since it's not a simple counting query) but
        # still satisfies the Above Threshold condition of the queries needing
        # sensitivity of 1 .

    return q


def pick_b(epsilon):
    queries = [make_query(b) for b in range(1, 1000)]
    b = above_threshold(queries, adult, T=0, epsilon=epsilon)
    return b


pick_b(1.0)
```

```python
# TEST CASE
b = pick_b(1.0)
assert b > 80
assert b < 100
```

## Question 6

Implement `above_threshold_val`, which returns the _value_ of the first query result above the threshold. Your solution should have a **total privacy cost** of `epsilon`.


```python
def make_query(age):
    def q(df):
        return len(df[df["Age"] == age])

    return q


# Normally we don't need the actual query value returned, just the index.
# But incase we actually do want the value, this is how we would do it.
def above_threshold_val(queries, df, T, epsilon):
    idx = above_threshold(queries, df, T=T, epsilon=epsilon / 2)
    # The returned value needs noise added to it to satisfy DP!
    return laplace_mech(queries[idx](df), sensitivity=1, epsilon=epsilon / 2)


queries = [make_query(age) for age in range(0, 100)]
above_threshold_val(queries, adult, 800, 1.0)
```

```python
# TEST CASE
queries = [make_query(age) for age in range(0, 100)]

results = [above_threshold_val(queries, adult, 800, 1.0) for _ in range(20)]
assert np.mean(results) > 865
assert np.mean(results) < 890
```
