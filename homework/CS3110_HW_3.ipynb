```python
# Load the data and libraries
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from unittest.mock import patch


def laplace_mech(v, sensitivity, epsilon):
    return v + np.random.laplace(loc=0, scale=sensitivity / epsilon)


adult = pd.read_csv(
    "https://github.com/jnear/cs3110-data-privacy/raw/main/homework/adult_with_pii.csv"
)
adult = adult.dropna()
```

## Question 1 (10 points)

Write code to answer the query: "how many participants have never been married?"

_Hint_: filter the `adult_data` dataframe to contain only participants who were never married, then return the `len` of the filtered dataframe.


```python
# list all the possible values in the "Marital Status" column
# print(adult["Marital Status"].unique())
# Looking at the possible values, the only one we are counting is "Never-married"


def query1():
    return len(adult[adult["Marital Status"] == "Never-married"])


query1()
```

```python
# TEST CASE for question 1
assert query1() == 9726
```

## Question 2 (5 points)

In 2-5 sentences, answer the following:

- What is the sensitivity of `query1`, and why?


The sensitivity of this query is defined as the maximum difference in output
values when the query is applied over all sets of _neighboring_ datasets. In
this scenario, any neighboring dataset is one where a single row is either
_removed_ or _added_. Adding or removing a row from a dataset this query is
acting on will, at-most, increase or decrease the query result by one.
Therefore, **the sensitivity of query1 is 1**.


## Question 3 (10 points)

Use the implementation of `laplace_mech` to produce a differentially private answer to `query1`, with `epsilon = 0.1`.


```python
def dp_query1(epsilon):
    sensitivity = 1  # as figured out in the previous question
    return laplace_mech(query1(), sensitivity, epsilon)


dp_query1(0.1)
```

```python
# TEST CASE for question 3
dp_results = [dp_query1(0.1) for _ in range(100)]
spec = [np.random.laplace(loc=9726, scale=1 / 0.1) for _ in range(100)]
assert stats.wasserstein_distance(dp_results, spec) < 10
```

## Question 4 (10 points)

The `pct_error` function, defined below, returns the percent relative error between an original query result and a differentially private result for the same query.

Implement a function `graph_error1` that:

- Calculates 1000 differentially private answers to `dp_query1`
- Calculates the percent error for each one of these answers against the original (non-private) answer
- Graphs the distribution of errors using a histogram

_Hint_: use `plt.hist(..., bins=20)`.

The given code will use your function to plot errors for `epsilon=0.1` and `epsilon=1.0`.


```python
def pct_error(orig, priv):
    return np.abs(orig - priv) / orig * 100.0
```

```python
def graph_error1(epsilon):
    # Calculate dp_query1 1000 times and save the results
    results = [dp_query1(epsilon) for _ in range(1000)]

    # Calculate the percent error using the original query1 result and the
    # function defined above
    percent_errors = pct_error(query1(), np.array(results))

    # Graph the distribution using a histogram
    plt.hist(percent_errors, bins=20)
    plt.xlabel("Percent Error")
    plt.ylabel("Frequency")
    # plt.show()


graph_error1(0.1)
graph_error1(1.0)
```

```python
error_spec = (
    100.0 * np.abs(np.random.laplace(loc=0, scale=1, size=1000)) / query1()
)

with patch("matplotlib.pyplot.hist") as mock_hist:
    graph_error1(1.0)

args, kwargs = mock_hist.call_args
assert stats.wasserstein_distance(error_spec, args[0]) < 0.001
```

## Question 6 (10 points)

Consider `query2`, which asks how many people in the dataset are over the age of 60.


```python
def query2():
    return len(adult[adult["Age"] > 60])
```

Implement `dp_query2`, a differentially private version of `query2` (as in question 3), and `graph_error2`, which graphs relative error for `dp_query2` (as in question 4).


```python
def dp_query2(epsilon):
    sensitivity = 1  # because this is a counting query
    return laplace_mech(query2(), sensitivity, epsilon)


def graph_error2(epsilon):
    # Calculate dp_query2 1000 times and save the results
    results = [dp_query2(epsilon) for _ in range(1000)]

    # Calculate the percent error using the original query2 result and the
    # function defined above
    percent_errors = pct_error(query2(), np.array(results))

    # Graph the distribution using a histogram
    plt.hist(percent_errors, bins=20)
    plt.xlabel("Percent Error")
    plt.ylabel("Frequency")


graph_error2(1.0)
graph_error1(
    1.0
)  # we plot both errors for query 1 and query 2 at the same epsilon, to compare
```

```python
error_spec = (
    100.0 * np.abs(np.random.laplace(loc=0, scale=1, size=1000)) / query2()
)

with patch("matplotlib.pyplot.hist") as mock_hist:
    graph_error2(1.0)

args, kwargs = mock_hist.call_args
assert stats.wasserstein_distance(error_spec, args[0]) < 0.01
```

- What property of the query causes the difference in relative errors between `dp_query1` and `dp_query2`?


- `dp_query1` and `dp_query2` show different relative error distributions, even
  when using the same value of $\epsilon$. `dp_query2` produces larger relative
  errors than `dp_query1`.
- The difference in relative errors is due to `query1` having a larger output than `query2` (9726 vs. 1806).
  The Laplace noise we introduce does not depend on the output value of the original
  queries, it adds a value with magnitude roughly equal to $1/\epsilon$. So if
  $\epsilon$ doesn't change between the two queries, then the magnitude of the
  noise doesn't change as well. Therefore, queries that produce larger outputs will have smaller
  relative errors than those that produce smaller outputs.
